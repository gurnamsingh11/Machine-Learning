```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import make_blobs
import pandas as pd

%matplotlib inline
```

<p>Below code is used to generate a synthetic dataset with <code>n_samples=1000</code> data points, which will be divided into <code>centers=3</code> clusters. The data has <code>n_features=2</code>, meaning it will be two-dimensional.</p>

<p>Here’s a breakdown of each part:</p>

<ul>
  <li><strong><code>make_blobs</code></strong>: This is a function from the <code>sklearn.datasets</code> module, used to generate a set of data points with specified characteristics. It's often used for creating a synthetic dataset for classification tasks, such as clustering or classification algorithms.</li>
  <li><strong><code>n_samples=1000</code></strong>: This specifies the number of data points (samples) to generate. In this case, there will be 1000 data points in total.</li>
  <li><strong><code>centers=3</code></strong>: This defines the number of distinct clusters (centers) that the data points will be divided into. Here, 1000 points will be distributed among 3 different centers (clusters), which are represented by the centroids of each cluster.</li>
  <li><strong><code>n_features=2</code></strong>: This sets the dimensionality of the data points. Since <code>n_features=2</code>, each data point will have 2 features, meaning the data will be represented in a 2D space.</li>
</ul>

<p>After calling <code>make_blobs()</code>, the function returns two variables:</p>

<ul>
  <li><strong><code>X</code></strong>: The features (coordinates) of the generated data points. It is a 2D array of shape <code>(1000, 2)</code> where each row is a 2D point (x, y).</li>
  <li><strong><code>y</code></strong>: The labels corresponding to each data point, indicating the cluster to which each point belongs. It's a 1D array of length 1000, containing the cluster label (0, 1, or 2) for each sample.</li>
</ul>

<p>In summary, this function call creates a 2D dataset with 1000 data points grouped into 3 clusters. The points are represented by their (x, y) coordinates, and the labels (<code>y</code>) indicate the cluster assignment of each point.</p>


```python
X, y = make_blobs(n_samples=1000, centers=3, n_features=2)
```

### Plot coordinates

<p>The notation <code>X[:, 0]</code> and <code>X[:, 1]</code> is used to index and extract specific columns from the 2D array <code>X</code>. Here's what it means:</p>

<h3><code>X[:, 0]</code></h3>

<ul>
  <li><code>X</code> is a 2D array (or matrix) where each row represents a data point, and each column represents a feature (or attribute).</li>
  <li><code>:</code> (colon) in the first part (<code>X[:, 0]</code>) means "select all rows." This indicates that we want to consider all the data points (all rows) in the matrix.</li>
  <li><code>0</code> refers to the first column (indexing starts from 0 in Python). So <code>X[:, 0]</code> will give you the first feature of every data point (all rows, first column).</li>
</ul>

<p>For example, if <code>X</code> is:</p>

<pre>
X = [[1.5, 2.3],
     [3.1, 4.4],
     [0.7, 1.8]]
</pre>

<p>Then:</p>
<ul>
  <li><code>X[:, 0]</code> would give you the first column (first feature) for all rows: <code>[1.5, 3.1, 0.7]</code></li>
</ul>

<h3><code>X[:, 1]</code></h3>

<ul>
  <li>Similarly, <code>X[:, 1]</code> selects all the rows and the second column (since indexing starts at 0).</li>
  <li>So, <code>X[:, 1]</code> will give you the second feature of every data point.</li>
</ul>

<p>For the same <code>X</code> example:</p>

<ul>
  <li><code>X[:, 1]</code> would give you the second column (second feature) for all rows: <code>[2.3, 4.4, 1.8]</code></li>
</ul>

<h3>Summary</h3>

<ul>
  <li><code>X[:, 0]</code> extracts all the data points' first features (i.e., the first column).</li>
  <li><code>X[:, 1]</code> extracts all the data points' second features (i.e., the second column).</li>
</ul>

<p>These operations are commonly used when you're working with multidimensional data, especially in tasks like plotting or processing individual features of a dataset.</p>


<h3>Key Components:</h3>
<ul>
  <li><strong><code>X[:, 0]</code> and <code>X[:, 1]</code>:</strong>
    <ul>
      <li><code>X[:, 0]</code> extracts the first feature (or the x-coordinates of the data points in 2D space).</li>
      <li><code>X[:, 1]</code> extracts the second feature (or the y-coordinates of the data points in 2D space).</li>
      <li>Together, <code>X[:, 0]</code> and <code>X[:, 1]</code> represent the 2D positions of the data points in the plot.</li>
    </ul>
  </li>
  
  <li><strong><code>c=y</code>:</strong>
    <ul>
      <li><code>y</code> is the array of labels generated by <code>make_blobs</code>. These labels represent the cluster assignments for each data point (for example, 0, 1, or 2).</li>
      <li>The <code>c=y</code> argument tells <code>plt.scatter()</code> to color the data points based on their corresponding cluster labels (<code>y</code>). Each cluster will have a different color in the plot.</li>
    </ul>
  </li>
</ul>

<h3>How Do We Get 3 Clusters?</h3>

<p>When you run <code>make_blobs(n_samples=1000, centers=3, n_features=2)</code>, it generates 1000 data points with 2 features (dimensions), but it groups these points into <strong>3 clusters</strong>.</p>

<ul>
  <li>The <code>centers=3</code> argument in <code>make_blobs()</code> means that the data will have 3 different centers (or clusters).</li>
  <li>Each data point is assigned to one of these 3 clusters.</li>
  <li>Even though the data is 2D (with two features), the clustering algorithm groups the points into 3 clusters based on their proximity to the centers.</li>
</ul>

<p>So, you have:</p>
<ul>
  <li>3 different cluster centers in a 2D space.</li>
  <li>1000 data points distributed among those 3 clusters.</li>
  <li><code>y</code> contains the cluster label for each point (0, 1, or 2).</li>
</ul>

<h3>Plotting:</h3>

<p>When you run <code>plt.scatter(X[:,0], X[:,1], c=y)</code>, the scatter plot:</p>
<ul>
  <li>Plots each data point based on its <code>X[:, 0]</code> (x-coordinate) and <code>X[:, 1]</code> (y-coordinate).</li>
  <li>Colors the points according to their cluster label (<code>y</code>), so each cluster is displayed in a different color.</li>
</ul>

<h3>Summary:</h3>
<ul>
  <li>The <strong>2 features</strong> define the 2D space for the data points.</li>
  <li>The <strong>3 clusters</strong> come from the <code>centers=3</code> argument, which defines 3 different cluster centers in the 2D space.</li>
  <li>The <strong><code>c=y</code></strong> argument colors the points based on their cluster label, so you can visually distinguish the 3 clusters in the plot.</li>
</ul>

<p>Even though there are only two features (2D data), the points are grouped into 3 clusters based on the centroids provided by <code>make_blobs</code>. The <code>plt.scatter</code> function just visualizes these clusters in 2D, with different colors for each cluster.</p>


```python
plt.scatter(X[:,0], X[:,1],c=y)
```

### Standardization

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.cluster import KMeans
```

#### Elbow method to select K Value

```python
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, init="k-means++")
    kmeans.fit(X_train_scaled)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

### Plot Elbow Curve

```python
plt.plot(range(1,11), wcss)
plt.xticks(range(1,11))
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, init="k-means++")
```

```python
kmeans.fit_predict(X_train_scaled)
```

```python
y_pred = kmeans.predict(X_test_scaled)
```

```python
plt.scatter(X_test[:,0], X_test[:,1],c=y_pred)
```

### Validating the k value 

- sometimes elbow method becomes difficult

1) kneelocator
2) Silhoutee scoring

#### kneelocator

```python
# pip install kneed
```

```python
from kneed import KneeLocator
```

```python
kl = KneeLocator(range(1,11), wcss, curve='convex', direction='decreasing')
```

```python
print(kl.elbow)
```

#### silhoutee scoring

```python
from sklearn.metrics import silhouette_score
```

```python
silhouetee_coeff = []

for k in range(2,11):
    kmeans = KMeans(n_clusters=k, init="k-means++")
    kmeans.fit(X_train_scaled)
    score = silhouette_score(X_train_scaled, kmeans.labels_)
    silhouetee_coeff.append(score)
```

```python
plt.plot(range(2,11), silhouetee_coeff)
plt.xticks(range(2,11))
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Coefficient")
plt.show()
```
