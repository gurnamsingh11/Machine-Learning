```python
corpus = """Hello Welcome, to Krish Naik's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""
```

```python
print(corpus)
```

### Paragraph --> Sentences

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize
```

```python
sent_tokenize(corpus) # !, . is separated to columns
```

### Paragraph/ Sentence --> Words

```python
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

<ul>
  <li>In <code>word_tokenize</code>, the word <code>'s</code> is included as one element, whereas in <code>wordpunct_tokenize</code>, it is split into two elements: <code>'</code> and <code>s</code>.</li>

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```

<ul>
  <li>In above, <code>'s</code> is still treated as one token.</li>
  <li><code>Tutorials.</code> will be treated as <code>Single Word</code>as it is in the middle</li>
  <li><code>NLP.</code> is the last word so it will treated separately</li>
</ul>
