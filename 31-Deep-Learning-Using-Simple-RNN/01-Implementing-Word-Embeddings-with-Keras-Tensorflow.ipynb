```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent = ['the glass of milk',
        'the glass of juice',
        'the cup of tea',
        'I am a good boy',
        'I am a good developer',
        'understand the meaning of words',
        'your videos are good']

len(sent)
```

### Define the vocabulary size

```python
voc_size = 10000
```

#### One Hot Representation

```python
one_hot_repr = [one_hot(words, voc_size) for words in sent]
one_hot_repr
```

```python
len(one_hot_repr)

# these are the indexes of each word where they are present in the vocab
```

#### Word Embedding

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.models import Sequential
from tensorflow.keras.utils import pad_sequences
```

```python
import numpy as np
```

```python
# we need to sentence fixed length of sentence (each sentence should have that many words max)
sent_length = 8
```

so if the size of a sentence is 4 and we have set length as 8 then pad sequences will add 4 zeros either at start or end

```python
embedded_docs = pad_sequences(one_hot_repr, padding='pre', maxlen=sent_length)
```

```python
embedded_docs
```

```python
#### Feature Representation
```

```python
dim = 10 # use 10 dimensions
```

```python
model = Sequential()
model.add(Embedding(voc_size, 
                    dim,
                    input_length = sent_length))
model.compile('adam', 'mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```

Each word is represented by 10 dimensions
